```python
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output
from html import escape
import base64

# ============================================================
# 🎀 STUDENT RECORDS
# ============================================================

students = {
    "Althea Angelie P. Intrepido": {
        "course": "BSCS", "year": "1",
        "subjects": [
            ["CSCC 101", "Introduction to Computing", "Brenda Balala", 5, 2],
            ["CSCC 102", "Fundamentals of Programming", "Arianne Joy Cari-an", 5, 2],
            ["CSMATH 1", "Linear Algebra", "Vic Ramos Tobias", 3, 2.25],
            ["WKL 1", "Introduksiyon sa Filipinolohiya", "Karen Umadhay", 3, 1.5],
            ["MST 111", "Mathematics in the Modern World", "Natalie Gallaron", 3, 2],
        ]
    },
    "Jayson Clark F. Plasquita": {
        "course": "BSHM", "year": "1",
        "subjects": [
            ["Euthenics 1", "Euthenics 1", "RHODA S. HERIDA", 3, 2],
            ["RM", "Risk Management as Applied to Safety, Security, and Sanitation", "FAITH SOLEJON", 3, 2],
            ["MPTH", "Macro Perspective of Tourism and Hospitality", "Donasco, Jimmy D.", 3, 1],
            ["RPH", "Readings in Philippine History", "Llamo, Robie Lyn A.", 3, 1.5],
            ["PATHFIT 1", "PATHFIT 1", "PAGUNTALAN, DAISY BELL S.", 3, 1],
        ]
    },
    "Kylie Joy S. Castro": {
        "course": "BSIT", "year": "1",
        "subjects": [
            ["GE201", "General Education 201", "Joyce Enano", 3, 1],
            ["GE204", "General Education 204", "Rose Mari Besonia", 3, 2],
            ["GE200", "General Education 200", "Mercedes Ramoran", 3, 1],
            ["LIT1", "Literature 1", "Sherly Mae Roda", 3, 1.5],
            ["CC101", "Computer Concepts 101", "Jane Alferes", 3, 2],
        ]
    },
    "Paul Daniel L. Intrepido": {
        "course": "BSCOM ENG", "year": "3",
        "subjects": [
            ["EP", "Engineering Physics", "Joseph Santos", 3, 1],
            ["DE", "Differential Equation", "Rebecca Flores", 3, 2],
            ["CAL", "Calculus", "Mark Bautista", 3, 2],
            ["EE", "Electronics Engineering", "Christian Ignacio", 3, 2],
            ["CE", "Circuits Engineering", "Elena Torres", 3, 1],
        ]
    },
    "Angelu Jhon D. Lumawag": {
        "course": "BSCM", "year": "1",
        "subjects": [
            ["ECL", "Engineering Chemistry Lecture", "Roxxane Joy Panaguiton", 3, 1],
            ["EM", "Engineering Mathematics 111", "Elmer Claveria", 3, 1],
            ["CPEPC 113", "CPEPC 113", "Christian Ignacio", 3, 1],
            ["CPEPC 111", "CPEPC 111", "Jenith Guaro", 3, 2],
            ["ECL", "Engineering Chemistry Laboratory", "Rebecca Flores", 3, 2],
        ]
    },
    "Riona Baylosis": {
        "course": "BSAG", "year": "1",
        "subjects": [
            ["Ansci1", "Animal Science 1", "Angela Cruz", 3, 1],
            ["Agri1", "Agriculture 1", "Mark Flores", 3, 2],
            ["SoilSci1", "Soil Science 1", "Sophia Reyes", 3, 1],
            ["CropProtection", "Crop Protection", "Carla Mendoza", 3, 1.5],
            ["CropSci", "Crop Science", "Peter Garcia", 3, 2.25],
        ]
    },
    "Theodora Alexes Laspinas": {
        "course": "BSARC", "year": "3",
        "subjects": [
            ["ARCH131", "Architecture 131", "Robert Lim", 5, 2],
            ["ARCH135", "Architecture 135", "Daniel Ramos", 5, 1.25],
            ["ARCH137", "Architecture 137", "Grace Villanueva", 3, 2],
            ["ARCH133", "Architecture 133", "Anthony Rivera", 3, 1.5],
            ["ARCH139", "Architecture 139", "Michelle Fernandez", 4, 1],
        ]
    },
    "Joshua Miguel T. Aquino": {
        "course": "BSCRIM", "year": "1",
        "subjects": [
            ["CRIM 101", "Criminology 101", "Eduardo Mercado", 3, 1.75],
            ["CRIM 102", "Criminology 102", "Rowena Salazar", 3, 2.25],
            ["CRIM 103", "Criminology 103", "Dennis Cabrera", 3, 2],
            ["GE 101", "General Education 101", "Maribel Ramos", 3, 1],
            ["PE 101", "Physical Education 101", "Jerome Padilla", 2, 1],
        ]
    },
    "Kevin Mark D. Fernandez": {
        "course": "BSED", "year": "1",
        "subjects": [
            ["EDU 101", "Education 101", "Alicia Manalo", 3, 1],
            ["ENG 101", "English 101", "Benjamin Soriano", 3, 2],
            ["EDU 102", "Education 102", "Felicia Dizon", 3, 3],
            ["GE 101", "General Education 101", "Ernesto Valdez", 3, 2],
            ["EDU 103", "Education 103", "Marissa Evangelista", 3, 1],
        ]
    },
    "Hannah Grace P. Morales": {
        "course": "BSN", "year": "1",
        "subjects": [
            ["NCM 101", "Nursing Care Management 101", "Patricia Villanueva", 5, 2],
            ["BIO 101", "Biology 101", "Samuel de Leon", 3, 2.25],
            ["BIO 102", "Biology 102", "Veronica Abad", 2, 1],
            ["CHEM 101", "Chemistry 101", "Christopher Pineda", 3, 1.25],
            ["NCM 102", "Nursing Care Management 102", "Rachelle Ong", 3, 2.25],
        ]
    }
}

# ============================================================
# 💜 PURPLE COQUETTE THEME
# ============================================================

display(HTML("""
<style>
.coquette {
    background: linear-gradient(135deg, #fffaff, #f0e2ff);
    border: 2px solid #c9a1e8;
    border-radius: 22px;
    padding: 24px;
    margin: 12px 0;
    box-shadow: 0 7px 20px #9b65b522;
    font-family: Arial, sans-serif;
    color: #513363;
}
.coquette-heading {
    text-align: center;
    color: #713b98;
    font-size: 24px;
    font-weight: bold;
    padding-bottom: 14px;
    border-bottom: 2px dashed #c49bdd;
}
.coquette-subtitle {
    text-align: center;
    color: #9864b5;
    margin: 12px 0 20px;
}
.report-title {
    font-weight: bold;
    color: #75439a;
    margin: 20px 0 10px;
    font-size: 16px;
}
.report-table {
    width: 100%;
    border-collapse: collapse;
    font-size: 12px;
}
.report-table th {
    background: #a979c7;
    color: white;
    padding: 10px 7px;
    text-align: left;
}
.report-table td {
    padding: 9px 7px;
    border-bottom: 1px solid #e9d9f3;
    overflow-wrap: anywhere;
}
.report-table tr:nth-child(even) {
    background: #f8f0ff;
}
.summary {
    background: #f8efff;
    border: 1px solid #dec4f0;
    border-radius: 14px;
    padding: 12px;
}
.summary table {
    width: 100%;
    border-collapse: collapse;
}
.summary td {
    padding: 9px 5px;
}
.summary td:last-child {
    text-align: right;
    font-weight: bold;
}
.gwa {
    background: #8148a8 !important;
    color: white !important;
    font-size: 15px;
}
.gwa td {
    color: white !important;
    padding: 13px 8px;
}
.coquette-footer {
    text-align: center;
    color: #9864b5;
    margin-top: 20px;
}
.jupyter-button {
    background: #9b62c4 !important;
    color: white !important;
    border: 2px solid #d8b9f2 !important;
    border-radius: 20px !important;
    font-weight: bold !important;
}
.jupyter-button:hover {
    background: #713b98 !important;
}
</style>
"""))

# ============================================================
# STUDENT DROPDOWN
# ============================================================

student_dropdown = widgets.Dropdown(
    options=list(students.keys()),
    value="Althea Angelie P. Intrepido",
    description="🎀 Choose Student:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="100%", max_width="650px")
)

report_output = widgets.Output()
action_output = widgets.Output()

# ============================================================
# REPORT GENERATOR
# ============================================================

def create_report_html(trigger_print=False):
    name = student_dropdown.value
    student = students[name]
    subjects = student["subjects"]

    total_subjects = len(subjects)
    total_units = sum(s[3] for s in subjects)
    weighted_total = sum(s[3] * s[4] for s in subjects)

    gwa = weighted_total / total_units if total_units else 0
    average = (
        sum(s[4] for s in subjects) / total_subjects
        if total_subjects else 0
    )

    rows = ""

    for code, description, teacher, units, grade in subjects:
        rows += f"""
        <tr>
            <td>{escape(str(code)) or "-"}</td>
            <td>{escape(str(description))}</td>
            <td>{escape(str(teacher))}</td>
            <td style="text-align:center">{units:g}</td>
            <td style="text-align:center">{grade:.2f}</td>
        </tr>
        """

    print_script = "<script>window.onload = function() { setTimeout(function() { window.print(); }, 500); }</script>" if trigger_print else ""

    return f"""
    <!DOCTYPE html>
    <html>
    <head>
    <meta charset="utf-8">
    <title>Student Grade Report - {escape(name)}</title>
    <style>
        body {{
            font-family: Arial, sans-serif;
            padding: 25px;
            color: #513363;
            background: white;
        }}
        .report {{
            border: 2px solid #c9a1e8;
            border-radius: 20px;
            padding: 22px;
            max-width: 800px;
            margin: 0 auto;
        }}
        h2 {{
            text-align: center;
            color: #713b98;
            border-bottom: 2px dashed #c49bdd;
            padding-bottom: 15px;
            margin-top: 0;
        }}
        .subtitle {{
            text-align: center;
            color: #9864b5;
        }}
        table {{
            width: 100%;
            border-collapse: collapse;
            margin-top: 15px;
            font-size: 12px;
        }}
        th {{
            background: #a979c7;
            color: white;
            padding: 9px;
            text-align: left;
        }}
        td {{
            border-bottom: 1px solid #e9d9f3;
            padding: 8px;
        }}
        tr:nth-child(even) {{
            background: #f8f0ff;
        }}
        .gwa {{
            background: #8148a8 !important;
            color: white;
            font-weight: bold;
        }}
        .gwa td {{
            color: white !important;
        }}
        .footer {{
            text-align: center;
            margin-top: 20px;
            color: #9864b5;
        }}
        @media print {{
            body {{
                padding: 0;
            }}
            .report {{
                border: none;
                padding: 0;
            }}
        }}
    </style>
    </head>
    <body>
    <div class="report">
        <h2>🎀 STUDENT GRADE REPORT 🎀</h2>
        <p class="subtitle">♡ Your academic progress, beautifully organized ♡</p>

        <p><b>Student Name:</b> {escape(name)}</p>
        <p><b>Course/Program:</b> {escape(student["course"])}</p>
        <p><b>Year Level:</b> {escape(student["year"])}</p>

        <h3>💜 Detailed Subject Grades</h3>
        <table>
            <tr>
                <th>Subject Code</th><th>Description</th>
                <th>Teacher</th><th>Units</th><th>Grade</th>
            </tr>
            {rows}
        </table>

        <h3>🎀 Grade Summary</h3>
        <table>
            <tr><td>Total Subjects Taken</td><td>{total_subjects}</td></tr>
            <tr><td>Total Units</td><td>{total_units:g}</td></tr>
            <tr><td>Simple Average Grade</td><td>{average:.2f}</td></tr>
            <tr class="gwa">
                <td>General Weighted Average (GWA)</td>
                <td>{gwa:.2f}</td>
            </tr>
        </table>

        <div class="footer">🎀 ♡ Keep shining and doing your best! ♡ 🎀</div>
    </div>
    {print_script}
    </body>
    </html>
    """


def show_report(change=None):
    with report_output:
        clear_output(wait=True)
        display(HTML(create_report_html(trigger_print=False)))


# ============================================================
# 🖨️ PRINT REPORT BUTTON
# ============================================================

print_button = widgets.Button(
    description="🖨️ Print Report",
    layout=widgets.Layout(width="190px", height="42px")
)


def print_report(button):
    html_report = create_report_html(trigger_print=True)
    encoded = base64.b64encode(
        html_report.encode("utf-8")
    ).decode("ascii")

    # Trigger direct window.open via injected JavaScript block
    with action_output:
        clear_output(wait=True)
        display(HTML(f"""
        <script>
            var win = window.open();
            if (win) {{
                win.document.write(atob('{encoded}'));
                win.document.close();
            }} else {{
                alert("Pop-up blocked! Please allow pop-ups for Colab to print automatically.");
            }}
        </script>
        """))


# ============================================================
# 💜 SAVE AS PDF BUTTON
# ============================================================

save_pdf_button = widgets.Button(
    description="💜 Save as PDF",
    layout=widgets.Layout(width="190px", height="42px")
)


def save_as_pdf(button):
    html_report = create_report_html(trigger_print=True)
    encoded = base64.b64encode(
        html_report.encode("utf-8")
    ).decode("ascii")

    # Trigger direct window.open via injected JavaScript block
    with action_output:
        clear_output(wait=True)
        display(HTML(f"""
        <script>
            var win = window.open();
            if (win) {{
                win.document.write(atob('{encoded}'));
                win.document.close();
            }} else {{
                alert("Pop-up blocked! Please allow pop-ups for Colab to save as PDF automatically.");
            }}
        </script>
        """))


# ============================================================
# CONNECT BUTTONS
# ============================================================

student_dropdown.observe(show_report, names="value")
print_button.on_click(print_report)
save_pdf_button.on_click(save_as_pdf)

# ============================================================
# DISPLAY THE APP
# ============================================================

display(HTML("""
<div class="coquette">
    <div class="coquette-heading">
        🎀 Student Grade Report Generator 🎀
    </div>
    <div class="coquette-subtitle">
        Select a student to view their grades 💜
    </div>
</div>
"""))

display(student_dropdown)

display(widgets.HBox(
    [print_button, save_pdf_button],
    layout=widgets.Layout(
        gap="12px",
        flex_flow="row wrap"
    )
))

display(report_output)
display(action_output)

# Show the first student's report initially
show_report()
```
