```python

import ipywidgets as widgets
from IPython.display import display, HTML, clear_output
from html import escape
import base64

# ============================================================
# 🎀 STUDENT RECORDS
# ============================================================

students = {
    "Althea Angelie P. Intrepido": {
        "course": "BSCS", "year": "1",
        "subjects": [
            ["CSCC 101", "Introduction to Computing", "Brenda Balala", 5, 1],
            ["CSCC 102", "Fundamentals of Programming", "Arianne Joy Cari-an", 5, 1],
            ["CSMATH 1", "Linear Algebra", "Vic Ramos Tobias", 3, 1],
            ["WKL 1", "Introduksiyon sa Filipinolohiya", "Karen Umadhay", 3, 1.5],
            ["MST 111", "Mathematics in the Modern World", "Natalie Gallaron", 3, 2],
        ]
    },
    "Jayson Clark F. Plasquita": {
        "course": "BSHM", "year": "1",
        "subjects": [
            ["Euthenics 1", "Euthenics 1", "Rhoda S. Herid", 3, 1.50],
            ["RM", "Risk Management as Applied to Safety, Security, and Sanitation", "Faith Ssolejon", 3, 2],
            ["MPTH", "Macro Perspective of Tourism and Hospitality", "Donasco, Jimmy D.", 3, 1],
            ["RPH", "Readings in Philippine History", "Llamo, Robie Lyn A.", 3, 1.5],
            ["PATHFIT 1", "PATHFIT 1", "Paguntalan, Daisy bell", 3, 1],
        ]
    },
    "Kylie Joy S. Castro": {
        "course": "BSIT", "year": "1",
        "subjects": [
            ["GE201", "General Education 201", "Joyce Enano", 3, 1],
            ["GE204", "General Education 204", "Rose Mari Besonia", 3, 2],
            ["GE200", "General Education 200", "Mercedes Ramoran", 3, 1],
            ["LIT1", "Literature 1", "Sherly Mae Roda", 3, 1.5],
            ["CC101", "Computer Concepts 101", "Jane Alferes", 3, 2],
        ]
    },
    "Paul Daniel L. Intrepido": {
        "course": "BSCOM ENG", "year": "3",
        "subjects": [
            ["EP", "Engineering Physics", "Joseph Santos", 3, 1],
            ["DE", "Differential Equation", "Rebecca Flores", 3, 2],
            ["CAL", "Calculus", "Mark Bautista", 3, 2],
            ["EE", "Electronics Engineering", "Christian Ignacio", 3, 2],
            ["CE", "Circuits Engineering", "Elena Torres", 3, 1],
        ]
    },
    "Angelu Jhon D. Lumawag": {
        "course": "BSCM", "year": "1",
        "subjects": [
            ["ECL", "Engineering Chemistry Lecture", "Roxxane Joy Panaguiton", 3, 1],
            ["EM", "Engineering Mathematics 111", "Elmer Claveria", 3, 1],
            ["CPEPC 113", "CPEPC 113", "Christian Ignacio", 3, 1],
            ["CPEPC 111", "CPEPC 111", "Jenith Guaro", 3, 2],
            ["ECL", "Engineering Chemistry Laboratory", "Rebecca Flores", 3, 2],
        ]
    },
    "Riona Baylosis": {
        "course": "BSAG", "year": "1",
        "subjects": [
            ["Ansci1", "Animal Science 1", "Angela Cruz", 3, 1],
            ["Agri1", "Agriculture 1", "Mark Flores", 3, 2],
            ["SoilSci1", "Soil Science 1", "Sophia Reyes", 3, 1],
            ["CropProtection", "Crop Protection", "Carla Mendoza", 3, 1.5],
            ["CropSci", "Crop Science", "Peter Garcia", 3, 2.25],
        ]
    },
    "Theodora Alexes Laspinas": {
        "course": "BSARC", "year": "3",
        "subjects": [
            ["ARCH131", "Architecture 131", "Robert Lim", 5, 2],
            ["ARCH135", "Architecture 135", "Daniel Ramos", 5, 1.25],
            ["ARCH137", "Architecture 137", "Grace Villanueva", 3, 2],
            ["ARCH133", "Architecture 133", "Anthony Rivera", 3, 1.5],
            ["ARCH139", "Architecture 139", "Michelle Fernandez", 4, 1],
        ]
    },
    "Joshua Miguel T. Aquino": {
        "course": "BSCRIM", "year": "1",
        "subjects": [
            ["CRIM 101", "Criminology 101", "Eduardo Mercado", 3, 1.75],
            ["CRIM 102", "Criminology 102", "Rowena Salazar", 3, 2.25],
            ["CRIM 103", "Criminology 103", "Dennis Cabrera", 3, 2],
            ["GE 101", "General Education 101", "Maribel Ramos", 3, 1],
            ["PE 101", "Physical Education 101", "Jerome Padilla", 2, 1],
        ]
    },
    "Kevin Mark D. Fernandez": {
        "course": "BSED", "year": "1",
        "subjects": [
            ["EDU 101", "Education 101", "Alicia Manalo", 3, 1],
            ["ENG 101", "English 101", "Benjamin Soriano", 3, 2],
            ["EDU 102", "Education 102", "Felicia Dizon", 3, 3],
            ["GE 101", "General Education 101", "Ernesto Valdez", 3, 2],
            ["EDU 103", "Education 103", "Marissa Evangelista", 3, 1],
        ]
    },
    "Hannah Grace P. Morales": {
        "course": "BSN", "year": "1",
        "subjects": [
            ["NCM 101", "Nursing Care Management 101", "Patricia Villanueva", 5, 2],
            ["BIO 101", "Biology 101", "Samuel de Leon", 3, 2.25],
            ["BIO 102", "Biology 102", "Veronica Abad", 2, 1],
            ["CHEM 101", "Chemistry 101", "Christopher Pineda", 3, 1.25],
            ["NCM 102", "Nursing Care Management 102", "Rachelle Ong", 3, 2.25],
        ]
    }
}

# ============================================================
# 💜 PURPLE COQUETTE THEME
# ============================================================

display(HTML("""
<style>
.coquette {
    background: linear-gradient(135deg, #fffaff, #f0e2ff);
    border: 2px solid #c9a1e8;
    border-radius: 22px;
    padding: 24px;
    margin: 12px 0;
    box-shadow: 0 7px 20px #9b65b522;
    font-family: Arial, sans-serif;
    color: #513363;
}
.coquette-heading {
    text-align: center;
    color: #713b98;
    font-size: 24px;
    font-weight: bold;
    padding-bottom: 14px;
    border-bottom: 2px dashed #c49bdd;
}
.coquette-subtitle {
    text-align: center;
    color: #9864b5;
    margin: 12px 0 20px;
}
.report-table {
    width: 100%;
    border-collapse: collapse;
    font-size: 12px;
}
.report-table th {
    background: #a979c7;
    color: white;
    padding: 10px 7px;
    text-align: left;
}
.report-table td {
    padding: 9px 7px;
    border-bottom: 1px solid #e9d9f3;
    overflow-wrap: anywhere;
}
.report-table tr:nth-child(even) {
    background: #f8f0ff;
}
.coquette-footer {
    text-align: center;
    color: #9864b5;
    margin-top: 20px;
}
.jupyter-button {
    background: #9b62c4 !important;
    color: white !important;
    border: 2px solid #d8b9f2 !important;
    border-radius: 20px !important;
    font-weight: bold !important;
}
.jupyter-button:hover {
    background: #713b98 !important;
}
.rank-first { background: #fce8a8 !important; font-weight: bold; }
.rank-second { background: #eeeeF5 !important; font-weight: bold; }
.rank-third { background: #f8d9c8 !important; font-weight: bold; }
</style>
"""))

# ============================================================
# 🏆 RANKING CALCULATIONS
# ============================================================

def calculate_gwa(student):
    subjects = student["subjects"]
    total_units = sum(subject[3] for subject in subjects)

    if not total_units:
        return 0

    return sum(
        subject[3] * subject[4] for subject in subjects
    ) / total_units


def get_student_rankings():
    ranking_list = []

    for name, student in students.items():
        ranking_list.append({
            "name": name,
            "course": student["course"],
            "year": student["year"],
            "gwa": calculate_gwa(student)
        })

    # Lower GWA is better
    ranking_list.sort(key=lambda item: (item["gwa"], item["name"].lower()))

    previous_gwa = None
    current_rank = 0

    for position, item in enumerate(ranking_list, start=1):
        if previous_gwa is None or abs(item["gwa"] - previous_gwa) > 1e-9:
            current_rank = position

        item["rank"] = current_rank
        previous_gwa = item["gwa"]

    return ranking_list


# ============================================================
# STUDENT DROPDOWN AND OUTPUTS
# ============================================================

student_dropdown = widgets.Dropdown(
    options=list(students.keys()),
    value="Althea Angelie P. Intrepido",
    description="🎀 Choose Student:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="100%", max_width="650px")
)

report_output = widgets.Output()
action_output = widgets.Output()
ranking_output = widgets.Output()

# ============================================================
# EMOJI-SAFE PRINT WINDOW
# ============================================================

def open_print_window(html_report):
    # Encode the complete UTF-8 HTML as Base64.
    encoded = base64.b64encode(
        html_report.encode("utf-8")
    ).decode("ascii")

    # TextDecoder restores the original UTF-8 bytes and emoji.
    javascript = """
    <script>
    (function() {
        const encoded = "ENCODED_HTML";
        const binary = atob(encoded);
        const bytes = Uint8Array.from(
            binary, character => character.charCodeAt(0)
        );
        const html = new TextDecoder("utf-8").decode(bytes);

        const win = window.open("", "_blank");

        if (!win) {
            alert("Pop-up blocked! Please allow pop-ups for this notebook.");
            return;
        }

        win.document.open();
        win.document.write(html);
        win.document.close();
    })();
    </script>
    """.replace("ENCODED_HTML", encoded)

    with action_output:
        clear_output(wait=True)
        display(HTML(javascript))


# ============================================================
# REPORT GENERATOR
# ============================================================

def create_report_html(trigger_print=False):
    name = student_dropdown.value
    student = students[name]
    subjects = student["subjects"]

    total_subjects = len(subjects)
    total_units = sum(s[3] for s in subjects)
    weighted_total = sum(s[3] * s[4] for s in subjects)

    gwa = weighted_total / total_units if total_units else 0
    average = (
        sum(s[4] for s in subjects) / total_subjects
        if total_subjects else 0
    )

    student_rank = next(
        item["rank"] for item in get_student_rankings()
        if item["name"] == name
    )

    rows = ""
    for code, description, teacher, units, grade in subjects:
        rows += f"""
        <tr>
            <td>{escape(str(code))}</td>
            <td>{escape(str(description))}</td>
            <td>{escape(str(teacher))}</td>
            <td style="text-align:center">{units:g}</td>
            <td style="text-align:center">{grade:.2f}</td>
        </tr>
        """

    print_script = """
    <script>
    window.addEventListener('load', function() {
        setTimeout(function() {
            window.focus();
            window.print();
        }, 1000);
    });
    </script>
    """ if trigger_print else ""

    return f"""<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1">
<title>Student Grade Report - {escape(name)}</title>
<style>
    body {{
        font-family: Arial, "Segoe UI Emoji", "Apple Color Emoji",
                     "Noto Color Emoji", sans-serif;
        padding: 25px;
        color: #513363;
        background: white;
    }}
    .emoji {{
        font-family: "Segoe UI Emoji", "Apple Color Emoji",
                     "Noto Color Emoji", sans-serif;
    }}
    .report {{
        border: 2px solid #c9a1e8;
        border-radius: 20px;
        padding: 22px;
        max-width: 800px;
        margin: 0 auto;
    }}
    h2 {{
        text-align: center;
        color: #713b98;
        border-bottom: 2px dashed #c49bdd;
        padding-bottom: 15px;
    }}
    h3 {{ color: #75439a; }}
    .subtitle, .footer {{
        text-align: center;
        color: #9864b5;
    }}
    table {{
        width: 100%;
        border-collapse: collapse;
        margin-top: 15px;
        font-size: 12px;
    }}
    th {{
        background: #a979c7;
        color: white;
        padding: 9px;
        text-align: left;
        -webkit-print-color-adjust: exact;
        print-color-adjust: exact;
    }}
    td {{
        border-bottom: 1px solid #e9d9f3;
        padding: 8px;
    }}
    tr:nth-child(even) {{
        background: #f8f0ff;
        -webkit-print-color-adjust: exact;
        print-color-adjust: exact;
    }}
    .gwa {{
        background: #8148a8 !important;
        color: white !important;
        font-weight: bold;
        -webkit-print-color-adjust: exact;
        print-color-adjust: exact;
    }}
    .gwa td {{ color: white !important; }}
    .footer {{ margin-top: 20px; }}
    @page {{ size: auto; margin: 12mm; }}
    @media print {{
        body {{ padding: 0; }}
        .report {{
            border: none;
            padding: 0;
            max-width: none;
        }}
        thead {{ display: table-header-group; }}
        tr {{ break-inside: avoid; }}
    }}
</style>
</head>
<body>
<div class="report">
    <h2><span class="emoji">🎀</span> STUDENT GRADE REPORT
        <span class="emoji">🎀</span></h2>
    <p class="subtitle">♡ Your academic progress, beautifully organized ♡</p>

    <p><b>Student Name:</b> {escape(name)}</p>
    <p><b>Course/Program:</b> {escape(student["course"])}</p>
    <p><b>Year Level:</b> {escape(student["year"])}</p>
    <p><b>Overall Student Rank:</b> #{student_rank} out of {len(students)}</p>

    <h3><span class="emoji">💜</span> Detailed Subject Grades</h3>
    <table>
        <thead>
        <tr>
            <th>Subject Code</th><th>Description</th>
            <th>Teacher</th><th>Units</th><th>Grade</th>
        </tr>
        </thead>
        <tbody>{rows}</tbody>
    </table>

    <h3><span class="emoji">🎀</span> Grade Summary</h3>
    <table>
        <tr><td>Total Subjects Taken</td><td>{total_subjects}</td></tr>
        <tr><td>Total Units</td><td>{total_units:g}</td></tr>
        <tr><td>Simple Average Grade</td><td>{average:.2f}</td></tr>
        <tr class="gwa">
            <td>General Weighted Average (GWA)</td>
            <td>{gwa:.2f}</td>
        </tr>
    </table>

    <div class="footer">
        <span class="emoji">🎀</span>
        ♡ Keep shining and doing your best! ♡
        <span class="emoji">🎀</span>
    </div>
</div>
{print_script}
</body>
</html>"""


def show_report(change=None):
    with report_output:
        clear_output(wait=True)
        display(HTML(create_report_html()))


# ============================================================
# 🏆 COMPLETE STUDENT RANKING TABLE
# ============================================================

def create_ranking_html():
    rankings = get_student_rankings()
    rows = ""

    for student in rankings:
        rank = student["rank"]

        if rank == 1:
            medal, row_class = "🥇", "rank-first"
        elif rank == 2:
            medal, row_class = "🥈", "rank-second"
        elif rank == 3:
            medal, row_class = "🥉", "rank-third"
        else:
            medal, row_class = "🎀", ""

        rows += f"""
        <tr class="{row_class}">
            <td style="text-align:center">{medal} {rank}</td>
            <td>{escape(student["name"])}</td>
            <td>{escape(student["course"])}</td>
            <td style="text-align:center">{student["gwa"]:.2f}</td>
        </tr>
        """

    return f"""
    <div class="coquette">
        <div class="coquette-heading">
            🏆 Student Grade Rankings 🏆
        </div>
        <div class="coquette-subtitle">
            ♡ Academic Ranking of All Students ♡<br>
            Lower GWA = Higher Ranking
        </div>
        <table class="report-table">
            <thead>
                <tr>
                    <th style="text-align:center">Rank</th>
                    <th>Student Name</th>
                    <th>Course</th>
                    <th style="text-align:center">GWA</th>
                </tr>
            </thead>
            <tbody>{rows}</tbody>
        </table>
        <div class="coquette-footer">
            🎀 🥇 Top 3 Students Are Highlighted 🥇 🎀
        </div>
    </div>
    """


def show_rankings(change=None):
    with ranking_output:
        clear_output(wait=True)
        display(HTML(create_ranking_html()))


# ============================================================
# 🖨️ PRINT REPORT BUTTON
# ============================================================

print_button = widgets.Button(
    description="🖨️ Print Report",
    layout=widgets.Layout(width="190px", height="42px")
)


def print_report(button):
    open_print_window(create_report_html(trigger_print=True))


# ============================================================
# 💜 SAVE AS PDF BUTTON
# ============================================================

save_pdf_button = widgets.Button(
    description="💜 Save as PDF",
    layout=widgets.Layout(width="190px", height="42px")
)


def save_as_pdf(button):
    # Opens the report's print dialog.
    # Select "Save as PDF" in the printer destination.
    open_print_window(create_report_html(trigger_print=True))


# ============================================================
# CONNECT BUTTONS AND DROPDOWN
# ============================================================

student_dropdown.observe(show_report, names="value")
print_button.on_click(print_report)
save_pdf_button.on_click(save_as_pdf)

# ============================================================
# 💜 DISPLAY THE COMPLETE APP
# ============================================================

display(HTML("""
<div class="coquette">
    <div class="coquette-heading">
        🎀 Student Grade Report Generator 🎀
    </div>
    <div class="coquette-subtitle">
        Select a student to view their grades 💜
    </div>
</div>
"""))

display(student_dropdown)

display(widgets.HBox(
    [print_button, save_pdf_button],
    layout=widgets.Layout(gap="12px", flex_flow="row wrap")
))

display(report_output)
display(action_output)
display(ranking_output)

show_report()
show_rankings()
```
